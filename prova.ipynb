QUESTÃO 02 – Soma dos Números (2 pontos)
---------------------------------------------------------

Solicite ao usuário um número inteiro positivo.

Utilizando um laço de repetição, calcule e exiba a soma de
todos os números de 1 até o número informado.

Exemplo:

Digite um número: 5

Resultado: 15

Pois:

1 + 2 + 3 + 4 + 5 = 15

```python
while True:
    n1 = int(input("Digite um numero inteiro positivo: "))
    if n1 > 0:
        break
    else:
        print("Por favor, digite um número inteiro positivo.")

soma = 0
sum_str = ""
for i in range (1,n1+1):
  soma = soma + i
  sum_str += str(i)
  if i < n1:
    sum_str += " + "
print(f'Resultado: {soma}')
print(f'Pois: {sum_str} = {soma}')
```

QUESTÃO 04 – Fatorial (4 pontos)
---------------------------------------------------------

Solicite ao usuário um número inteiro positivo e calcule
seu fatorial utilizando um laço de repetição.

Exemplo:

Digite um número: 5

Fatorial = 120

Pois:

5! = 5 × 4 × 3 × 2 × 1

Observação:
O programa deve validar se o número informado é positivo.

```python
numb = int  (input("Digite um numero inteiro positivo: "))
while numb < 0:
  numb = int  (input("Por favor, digite um numero inteiro positivo: "))
fatorial = 1
for i in range (1,numb+1):
  fatorial = fatorial * i
print(f'Fatorial = {fatorial}')
print('Pois:')
print(f'{numb}! = ',end='')
for i in range (numb,1,-1):
  print(f'{i} × ',end='')
print('1')
```
