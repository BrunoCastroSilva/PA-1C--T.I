```python
# lista vazia de numeros
# pedir 10 numeros
# lista impar e outra par
# separe os valores em cada lista

numb = []
impar = []
par = []
for i in range(10):
  numb.append(int(input("Digite um numero: ")))
  if numb[i] % 2 == 0:
    par.append(numb[i])
  else:
    impar.append(numb[i])
print('lista ', numb)
print('pares ', par)
print('impares ', impar)
```

```python

# sistema qyue guarda e mostra notas de 5 alunos em uma lista # mostre todas
# mostre a maior # media da turma
# quantidade de alunos com media maior ou igual a 6
# quantidade de alunos com media menor ou igual a 6

notas = []
for i in range(1,6):
  nota = float(input('Digite a nota do aluno: '))
  notas.append(nota)

maior6 = len([n for n in notas if n >= 6])
menor6 = len([n for n in notas if n < 6])

print(f'Todas as notas: {notas}')
print(f'Maior nota: {max(notas)}')
media_turma = sum(notas)/len(notas)
print(f'Média da turma: {media_turma:.2f}')
print(f'Quantidade de alunos com média maior ou igual a 6: {maior6}')
print(f'Quantidade de alunos com média menor que 6: {menor6}')
```
