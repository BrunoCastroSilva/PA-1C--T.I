```python
#10

height = int(input("Enter the height of the pyramid: "))

for i in range(1, height + 1):
    # Build the list of number strings first, then concatenate manually
    temp_numbers = []
    for j in range(i, 0, -1): # Decreasing sequence
        temp_numbers.append(str(j))
    for j in range(2, i + 1): # Increasing sequence (starts from 2)
        temp_numbers.append(str(j))

    # Manually concatenate the numbers with spaces, without using .join()
    numbers_str = ""
    for idx, num_str in enumerate(temp_numbers):
        numbers_str += num_str
        if idx < len(temp_numbers) - 1:
            numbers_str += " "

    # Calculate leading spaces for centering. Each number and its space take roughly 2 characters.
    leading_spaces = " " * (2 * (height - i))

    print(leading_spaces + numbers_str)
```
