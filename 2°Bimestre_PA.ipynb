```python
#10

height = int(input("Enter the height of the pyramid: "))

for i in range(1, height + 1):
    # Build the list of number strings first, then concatenate manually
    temp_numbers = []
    for j in range(i, 0, -1): # Decreasing sequence
        temp_numbers.append(str(j))
    for j in range(2, i + 1): # Increasing sequence (starts from 2)
        temp_numbers.append(str(j))

    # Manually concatenate the numbers with spaces, without using .join()
    numbers_str = ""
    for idx, num_str in enumerate(temp_numbers):
        numbers_str += num_str
        if idx < len(temp_numbers) - 1:
            numbers_str += " "

    # Calculate leading spaces for centering. Each number and its space take roughly 2 characters.
    leading_spaces = " " * (2 * (height - i))

    print(leading_spaces + numbers_str)
```

```python
while True:
    try:
        numero = int(input("Digite um número inteiro positivo: "))

        if numero < 0:
            print("Digite apenas números positivos.")
        else:
            break

    except:
        print("Entrada inválida!")

invertido = ""

while numero > 0:
    digito = numero % 10
    invertido += str(digito)
    numero = numero // 10

print("Número invertido:", invertido)
```

```python
while True:
    try:
        numero = int(input("Digite um número inteiro positivo: "))

        if numero < 0:
            print("Digite apenas números positivos.")
        else:
            break

    except:
        print("Entrada inválida!")

contador = 0

if numero == 0:
    contador = 1
else:
    while numero > 0:
        contador += 1
        numero = numero // 10

print("Quantidade de dígitos:", contador)
```

```python
meta = 10000
total = 0
mes = 0

while total < meta:

    while True:
        try:
            deposito = float(input("Valor do depósito do mês: "))

            if deposito > 0:
                break
            else:
                print("Digite um valor maior que zero.")

        except:
            print("Valor inválido!")

    total += deposito
    mes += 1

print("\nMeta atingida!")
print("Meses necessários:", mes)
print("Total acumulado: R$ {:.2f}".format(total))
print("Meta atingida no mês:", mes)
```

```python
subtotal = 0

while True:

    while True:
        try:
            preco = float(input("Preço do produto (0 para finalizar): "))
            break
        except:
            print("Valor inválido!")

    if preco == 0:
        break

    while True:
        try:
            quantidade = int(input("Quantidade: "))

            if quantidade > 0:
                break
            else:
                print("Digite uma quantidade maior que zero.")

        except:
            print("Valor inválido!")

    subtotal += preco * quantidade

if subtotal >= 1000:
    desconto = 15
elif subtotal >= 500:
    desconto = 10
elif subtotal >= 200:
    desconto = 5
else:
    desconto = 0

valor_desconto = subtotal * desconto / 100
valor_final = subtotal - valor_desconto

print("\nSubtotal: R$ {:.2f}".format(subtotal))
print("Desconto:", str(desconto) + "%")
print("Valor final: R$ {:.2f}".format(valor_final))
```
